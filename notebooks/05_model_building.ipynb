## section 1 

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

## Section 2 
load the final data set

```python
df = pd.read_csv("../data/final_uber_dataset.csv")
```

```python
df.columns
```

## Section 3 
choose x and y

```python
X = df.drop('fare_amount', axis=1)
Y = df['fare_amount']
```

```python
print(X.shape)
print(Y.shape)
```

## Section 4

train - test - split

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42
)
```

verify the spliting 

this means 80% of data is for training 

and the 20% of data is for testing


```python
print(X_train.shape)
print(X_test.shape)
print(Y_train.shape)
print(Y_test.shape)
```

## Section 5
Train the First Model 

The model will learn things like:

Distance ↑  → Fare ↑

Passengers ↑ → Fare may change

Time changes → Fare may change


```python
# select the linear model 
model = LinearRegression()
```

```python
# fit the linear model to the x andy training data 
model.fit(X_train, Y_train)
```

```python
# make new predictions on unseen data
Y_pred = model.predict(X_test)
```

## section 6 
make predictions

```python
predictions = pd.DataFrame({
    'Actual Fare': Y_test,
    'Predicted Fare': Y_pred
})

predictions.head(10)
```

```python
predictions.to_csv("../data/predictions.csv", index=False)
```
