# Findings

1. Dataset contains 200,000 Uber trips.
2. Target variable is fare_amount.
3. Dataset contains 6 useful predictive features.
4. pickup_datetime is stored as object and needs conversion.
5. dropoff_longitude contains 1 missing value.
6. dropoff_latitude contains 1 missing value.
7. key and Unnamed: 0 appear to be non-predictive columns.
8. Negative fare values exist and must be removed.
9. Passenger count contains invalid values such as 0 and 208.
10. Data cleaning is required before EDA and model building.
